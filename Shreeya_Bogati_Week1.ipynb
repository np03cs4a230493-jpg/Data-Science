```python
import numpy as np
```

```python
np.random.seed(42)
raw_temperatures = np.random.normal(loc=24.0, scale=3.0, size=72)

# overwrite 5 specific hours with unrealistic values, to simulate faulty sensor readings.
raw_temperatures[[5, 12, 28, 45, 60]] = [99.9, -15.0, 88.8, -50.0, 102.4]
```

```python
raw_temperatures
```

```python
in_range = (raw_temperatures >= 15.0) & (raw_temperatures <= 40.0)
```

```python
in_range
```

```python
# Counting how many readings passed vs failed the range check.
print("Realistic readings:", in_range.sum())
print("Faulty readings:", (~in_range).sum())
```

```python
# We take the median from ONLY the realistic readings (in_range selects just those).
# Using the median of everything would still be pulled off by the 5 broken values.
fallback_value = np.median(raw_temperatures[in_range])
fallback_value
```

```python
# np.where swaps in the fallback wherever a reading failed the check,
# and leaves every realistic reading untouched.
raw_temperatures = np.where(in_range, raw_temperatures, fallback_value)
raw_temperatures
```

```python
print("Mean:", round(np.mean(raw_temperatures), 2))
print("Standard Deviation:", round(np.std(raw_temperatures), 2))
print("Variance:", round(np.var(raw_temperatures), 2))
```

# Part 2: Sliding Median Filter

```python
# short signal with two obvious spikes: 99 and -45.
signal = np.array([10, 11, 99, 12, 13, -45, 14, 15, 16])
signal
```

```python
filtered_signal = signal.copy()
filtered_signal
```

```python
# Index 0 and the last index don't have a neighbour on both sides,
# so we only slide the window across positions 1 through 7.
for position in range(1, len(signal) - 1):
    neighbourhood = signal[position - 1 : position + 2]   # the value plus one neighbour on each side
    filtered_signal[position] = np.median(neighbourhood)   # the middle-ranked value replaces the centre

print("Original signal:", signal)
print("Filtered signal:", filtered_signal)
```
